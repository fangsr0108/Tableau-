```python
import pandas as pd
import numpy as np
```

```python
data_order = pd.read_excel('../data/项目1_commerce_order_2025.xlsx')
data_order
```

```python
# 利润四分位
Q1 = data_order['利润'].quantile(0.25)
Q3 = data_order['利润'].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR
```

```python
# 利润异常值
data_order['利润异常值标记'] = 0
data_order.loc[data_order['利润'] > upper_bound, '利润异常值标记'] = 1
data_order.loc[data_order['利润'] < lower_bound, '利润异常值标记'] = -1
data_order
```

```python
# 准时到货
data_order['是否超时'] = 0
# 超时到货
data_order.loc[data_order['实际发货时间'] > data_order['计划发货天数'], '是否超时'] = 1
# 提前到货
data_order.loc[data_order['实际发货时间'] < data_order['计划发货天数'], '是否超时'] = -1
data_order
```

```python
# 销售额四分位
Q1_SALES = data_order['销售额'].quantile(0.25)
Q3_SALES = data_order['销售额'].quantile(0.75)

IQR_SALES = Q3_SALES - Q1_SALES

lower_bound_SALES = Q1_SALES - 1.5 * IQR_SALES
upper_bound_SALES = Q3_SALES + 1.5 * IQR_SALES
# 销售额异常值
data_order['销售额异常值标记'] = 0
data_order.loc[data_order['销售额'] > upper_bound_SALES, '销售额异常值标记'] = 1
data_order.loc[data_order['销售额'] < lower_bound_SALES, '销售额异常值标记'] = -1
data_order
```

```python
data_order.to_csv('../data/commerce_order_alter.csv', index=False, encoding='utf-8-sig')
data_order
```
